## 2. Değişkenler

Değişken, bir değere verdiğimiz isimdir. `=` işareti "eşittir" değil, **"ata"** demektir.

```python
max_volume_hm3 = 306.65      # Tahtalı'nın normal kottaki hacmi (milyon m³)
current_volume_hm3 = 142.32  # 14 Ekim 2026 itibarıyla göldeki su
dam_name = "Tahtalı"         # metin (string) tırnak içinde yazılır

fill_ratio = current_volume_hm3 / max_volume_hm3
print(fill_ratio)
```

Python'da en sık kullanacağımız temel türler:

| Tür | Örnek | Anlamı |
|---|---|---|
| `int` | `2026` | tam sayı |
| `float` | `306.65` | ondalıklı sayı (Python'da ondalık ayracı **nokta**dır) |
| `str` | `"Tahtalı"` | metin |
| `bool` | `True` / `False` | doğru / yanlış |

`type()` ile bir değerin türünü öğrenebilirsin:

```python
print(type(max_volume_hm3), type(dam_name), type(2026))
```

## 3. f-string ile okunaklı çıktı

Bir metnin başına `f` koyarsan, süslü parantez `{}` içine değişken yazabilirsin. `:.1f` ise "virgülden sonra 1 basamak göster" demektir.

```python
print(f"{dam_name} Barajı %{fill_ratio * 100:.1f} dolu.")
```

## 4. Listeler

Liste, sıralı bir değerler topluluğudur. Aşağıda Tahtalı noktası için 2024 yılının aylık yağış toplamları var (mm, ERA5 verisi, yuvarlanmış).

```python
monthly_precip_2024 = [150, 50, 55, 42, 17, 0, 25, 26, 33, 1, 115, 238]
months = ["Oca", "Şub", "Mar", "Nis", "May", "Haz", "Tem", "Ağu", "Eyl", "Eki", "Kas", "Ara"]

print(len(monthly_precip_2024))   # kaç eleman var?
print(monthly_precip_2024[0])     # ilk eleman — Python saymaya 0'dan başlar!
print(monthly_precip_2024[-1])    # son eleman
print(monthly_precip_2024[5:8])   # 5, 6, 7. elemanlar (yaz ayları)
print(sum(monthly_precip_2024), max(monthly_precip_2024))
```

## 5. `for` döngüsü ve `if` koşulu

`for` döngüsü bir listedeki her eleman için aynı işi tekrarlar. **Girinti (4 boşluk) önemlidir:** döngünün içindeki satırlar girintili yazılır.

```python
for precip in monthly_precip_2024:
    if precip > 100:
        print(precip, "mm → çok yağışlı")
    elif precip < 5:
        print(precip, "mm → kurak")
    else:
        print(precip, "mm")
```

İki listeyi birlikte dolaşmak için `zip` kullanırız:

```python
for month, precip in zip(months, monthly_precip_2024):
    print(f"{month}: {precip} mm")
```

## 6. Sözlükler

Sözlük (`dict`), değerleri **isimle** saklar: `anahtar: değer`.

```python
tahtali = {
    "max_volume_hm3": 306.65,
    "dead_volume_hm3": 19.6,
    "lake_area_km2": 23.52,
    "catchment_area_km2": 546,
}
print(tahtali["lake_area_km2"])
tahtali["city"] = "İzmir"   # yeni anahtar ekleme
print(tahtali.keys())
```

## 7. Fonksiyonlar

Aynı hesabı defalarca yapacaksak onu bir **fonksiyon** hâline getiririz. `def` ile tanımlanır, `return` ile sonuç döndürür.

Örnek: Göle 1 mm yağmur yağarsa kaç milyon m³ su eklenir? 1 mm × 1 km² = 1000 m³ = 0,001 hm³.

```python
def rain_to_volume(precip_mm, area_km2):
    """Bir alana düşen yağışı hacme (hm³) çevirir."""
    return precip_mm * area_km2 * 0.001


print(rain_to_volume(100, 23.52))   # göle 100 mm düşerse
print(rain_to_volume(100, tahtali["catchment_area_km2"]))   # havzaya 100 mm düşerse
```
