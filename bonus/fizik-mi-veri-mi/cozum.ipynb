# Bonus — Fizik mi, Veri mi?

Üç gün boyunca bir **fiziksel model** kurduk: kütle korunumunu bir denkleme çevirdik, her terimin anlamını biliyorduk. Bugünlerde pek çok problem ise **makine öğrenmesi (ML)** ile çözülüyor: denklemi biz yazmıyoruz, model veriden kendisi öğreniyor.

Bu modülde aynı soruyu ML ile soracağız ve sonra **SHAP** ile ML modelinin "neye baktığını" açacağız.

**Ön koşul:** Gün 1 ve Gün 2.

> **Önemli bir dürüstlük notu:** Tahtalı'nın geçmiş doluluk serisi açık bir kaynakta yok. Bu yüzden ML modelini **gözlemlere değil, kendi fiziksel modelimizin çıktısına** eğiteceğiz. Yani ML modeli gerçeği değil, bizim kova modelimizi öğrenmeye çalışacak. Bu kurgu yine de çok öğretici: ML modelinin, bizim elle yazdığımız fiziği veriden **yeniden keşfedip keşfedemeyeceğini** görebiliriz. Gerçek bir çalışmada hedef değişken mutlaka gözlem olmalıdır.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from src.model import S0_DATE, S0_HM3, simulate

monthly = pd.read_csv("data/processed/tahtali_monthly.csv")
monthly.index = pd.to_datetime(monthly.pop("month"))
data = monthly.loc[S0_DATE:].copy()

physics = simulate(data["precipitation_mm"], data["et0_mm"], S0_HM3)
```

## 1. Hedef ve özellikler

- **Hedef (y):** Bir aydaki depolama değişimi, $\Delta S$ (hm³). Barajın o ay ne kadar dolduğu ya da boşaldığı.
- **Özellikler (X):** O ayın yağışı, ET0'ı ve takvim ayı (1–12).

Fiziksel modelimizde $\Delta S$ bu üç bilgiyle belirleniyor, **yeter ki** baraj taşmasın ya da ölü hacme inmesin. O eşiklerde sonuç depolamanın kendisine bağlı olur ve ML modeli depolamayı görmüyor. Temel parametrelerle 2019–2026 arasında baraj hiç taşmıyor ve ölü hacme inmiyor; eşiklerin etkisini "Zaman kalırsa" bölümünde göreceğiz.

```python
previous = physics["storage"].shift(1)
previous.iloc[0] = S0_HM3
data["delta_storage"] = physics["storage"] - previous
data["month_of_year"] = data.index.month

features = ["precipitation_mm", "et0_mm", "month_of_year"]
X, y = data[features], data["delta_storage"]
data.head()
```

## 2. Eğitim ve test verisini **zamana göre** ayırmak

ML'de modeli bir kısım veriyle eğitip, **hiç görmediği** başka bir kısımla test ederiz. Zaman serilerinde bu ayrım **rastgele yapılmaz**: rastgele ayırırsak model, test ayının bir ay öncesini ve sonrasını eğitimde görmüş olur. Bu, geleceği bilerek geçmişi tahmin etmek gibidir (*bilgi sızıntısı*). Doğrusu: geçmişle eğit, gelecekle test et.

```python
split = "2024-01-01"
X_train, X_test = X.loc[:split].iloc[:-1], X.loc[split:]
y_train, y_test = y.loc[:split].iloc[:-1], y.loc[split:]
print(f"Eğitim: {len(X_train)} ay ({X_train.index[0]:%Y-%m} → {X_train.index[-1]:%Y-%m})")
print(f"Test:   {len(X_test)} ay ({X_test.index[0]:%Y-%m} → {X_test.index[-1]:%Y-%m})")
```

## 3. Random Forest

**Random Forest**, çok sayıda karar ağacının ortalamasını alan, güçlü ve kolay kullanılan bir ML modelidir. scikit-learn'de her model aynı üç adımla çalışır: oluştur → `fit` → `predict`.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

model = RandomForestRegressor(n_estimators=300, random_state=42)
model.fit(X_train, y_train)
prediction = pd.Series(model.predict(X_test), index=X_test.index)

print(f"R²   = {r2_score(y_test, prediction):.2f}")
print(f"RMSE = {np.sqrt(mean_squared_error(y_test, prediction)):.2f} hm³/ay")
```

```python
fig, ax = plt.subplots(figsize=(10, 3.5))
ax.plot(y.index, y, color="black", label="Fiziksel model (hedef)")
ax.plot(prediction.index, prediction, color="tab:green", marker="o", markersize=3, label="Random Forest tahmini")
ax.axvline(pd.Timestamp(split), color="gray", linestyle="--")
ax.text(pd.Timestamp(split), ax.get_ylim()[1] * 0.9, "  test dönemi →", color="gray")
ax.axhline(0, color="black", linewidth=0.5)
ax.set_ylabel("ΔS (hm³/ay)")
ax.set_title("Aylık depolama değişimi: fiziksel model ve Random Forest")
ax.legend(loc="lower left")
plt.show()
```

### Alıştırma 1 (temel)
Random Forest hangi aylarda en çok yanılıyor? Test dönemindeki en büyük 3 hatayı bulun. O ayların yağışını, **eğitim** verisindeki yağışlarla karşılaştırın: eğitimdeki ayların yüzde kaçı o aydan daha az yağış almış?

```python
# BEGIN SOLUTION
errors = (prediction - y_test).abs().sort_values(ascending=False)
worst = errors.head(3).index
train_precip = X_train["precipitation_mm"]
print(pd.DataFrame({
    "hata (hm³)": errors.loc[worst],
    "yağış (mm)": X_test.loc[worst, "precipitation_mm"],
    "eğitimde daha az yağışlı ay (%)": [100 * (train_precip < p).mean() for p in X_test.loc[worst, "precipitation_mm"]],
}).round(1))
# END SOLUTION
```

En büyük hata, eğitim verisindeki ayların neredeyse hepsinden daha yağışlı bir ayda. Random Forest bir **ezberleyici** gibi çalışır: tahmini, eğitimde gördüğü benzer ayların ortalamasıdır. Eğitimde çok yağışlı ay az olduğu için uçlarda zorlanır, ve eğitimde görmediği bir aralığa **hiç çıkamaz** (dışdeğerleme yapamaz). Fiziksel modelde böyle bir sınır yok: denklem her yağış miktarı için geçerli.

## 4. SHAP: Model neye bakıyor?

ML modelleri çoğu zaman "kara kutu" olarak eleştirilir. **SHAP** (SHapley Additive exPlanations), her tahmini özelliklerin katkılarına ayırır: "Bu ay ΔS'nin tahmini +5 hm³; bunun +7'si yağıştan, −2'si ET0'dan geliyor" gibi.

```python
import shap

explainer = shap.TreeExplainer(model)
shap_values = explainer(X_test)

shap.plots.beeswarm(shap_values, show=False)
plt.title("Her özelliğin tahmine katkısı (test dönemi)")
plt.show()
```

**Nasıl okunur?**
- Her nokta bir ay. Yatay konum, o özelliğin o ayın tahminine katkısı (hm³).
- Renk, özelliğin değeri: kırmızı yüksek, mavi düşük.
- Özellikler, ortalama etkilerine göre yukarıdan aşağı sıralanır.

Beklentimiz: yağış en önemli özellik olmalı, ve yüksek yağış (kırmızı) pozitif katkı vermeli.

```python
shap.plots.scatter(shap_values[:, "precipitation_mm"], show=False)
plt.title("Yağışın katkısı yağışa nasıl bağlı?")
plt.show()
```

Fiziksel modelde yağışın etkisi **doğrusal**dı: $(C \cdot A_{havza} + A_{göl}) \approx 0{,}18$ hm³/mm. SHAP grafiğinde ilişki yaklaşık doğrusal mı? Basamaklı görünmesinin nedeni ne olabilir? (İpucu: karar ağaçları sürekli bir doğru değil, basamaklar çizer.)

### Alıştırma 2 (ileri)
Fiziksel modeldeki yağış katsayısını, $0{,}29 \cdot 546 \cdot 0{,}001 + 23{,}52 \cdot 0{,}001$ hm³/mm, hesaplayın. Sonra SHAP değerlerine bir doğru uydurarak (`np.polyfit`) ML modelinin "öğrendiği" eğimi bulun ve karşılaştırın.

```python
# BEGIN SOLUTION
physics_slope = 0.29 * 546 * 0.001 + 23.52 * 0.001
learned_slope = np.polyfit(X_test["precipitation_mm"], shap_values[:, "precipitation_mm"].values, 1)[0]
print(f"Fiziksel model: {physics_slope:.3f} hm³/mm, Random Forest (SHAP): {learned_slope:.3f} hm³/mm")
# END SOLUTION
```

### Alıştırma 3 (ileri): Eşikler ve depolama
1. Özelliklere bir önceki ayın sonundaki **depolamayı** (`previous`) ekleyin ve modeli yeniden eğitin. Test R² değeri iyileşiyor mu? Neden?
2. Şimdi barajın sık sık taştığı bir parametre setiyle ($C = 0{,}40$) hedefi yeniden üretin ve aynı karşılaştırmayı yapın: depolama olmadan ve depolamayla. Bu kez ne değişti?

```python
# BEGIN SOLUTION
from dataclasses import replace

from src.model import Params


def compare(runoff_coef):
    """Verilen C ile hedefi üretir; depolamasız ve depolamalı Random Forest'ın test R² değerlerini döndürür."""
    sim = simulate(data["precipitation_mm"], data["et0_mm"], S0_HM3, replace(Params(), runoff_coef=runoff_coef))
    prev = sim["storage"].shift(1)
    prev.iloc[0] = S0_HM3
    target = sim["storage"] - prev
    table = data[features].assign(storage_prev=prev)
    scores = {}
    for name, cols in [("depolamasız", features), ("depolamalı", features + ["storage_prev"])]:
        Xa, Xb = table.loc[:split, cols].iloc[:-1], table.loc[split:, cols]
        ya, yb = target.loc[:split].iloc[:-1], target.loc[split:]
        rf = RandomForestRegressor(n_estimators=300, random_state=42).fit(Xa, ya)
        scores[name] = round(r2_score(yb, rf.predict(Xb)), 2)
    scores["taşan ay sayısı"] = int((sim["spill"] > 0).sum())
    return scores


print(pd.DataFrame({f"C = {c}": compare(c) for c in [0.29, 0.40]}))
# END SOLUTION
```

Temel parametrelerle ($C = 0{,}29$) baraj hiç taşmadığı için $\Delta S$ depolamaya bağlı değil; depolamayı eklemek yalnızca gürültü ekler. $C = 0{,}40$ ile baraj sık sık taşar: depolamayı görmeyen model taşma aylarında büyük hatalar yapar ve depolamayı eklemek R² değerini belirgin biçimde artırır. **Bir ML modeli, ancak kendisine verilen bilgiyle öğrenebilir.**
